```python
import cv2
import numpy as np

def extract_features(img_path):
    img = cv2.imread(img_path, 0)
    img = cv2.resize(img, (28, 28))
    edges = cv2.Canny(img, 50, 150)
    f1=np.sum(edges>0)
    f2=np.sum(edges)
    f3=np.sum(img>200)

    ys, xs = np.where(edges > 0)

    if len(xs) > 0:
        min_x,max_x=np.min(xs), np.max(xs)
        min_y,max_y=np.min(ys), np.max(ys)
    else:
        min_x=max_x=min_y=max_y=0
    f4=len(xs)
    f5=max_x-min_x+1
    f6=max_y-min_y+1
    f7=(f5*100)/(f6+1e-5)
    f8=(f3*1000)/((f5*f6)+1e-5)

    return [f1,f2,f3,f4,f5,f6,f7,f8]
```

```python
features=extract_features("triangle.jpg")
print(features)
```

```python
def normalize(f):
    n1 = (f[0] - 113) / 22
    n2 = (f[1] - 28714) / 5564
    n3 = (f[2] - 161) / 82
    n4 = (f[3] - 15) / 3
    n5 = (f[4] - 14) / 3
    n6 = (f[5] - 107) / 10
    n7 = (f[6] - 150) / 35
    n8 = (f[7] - 737) / 195

    return [n1,n2,n3,n4,n5,n6,n7,n8]
```

```python
norm = normalize(features)
print(norm)
```

```python
def to_fixed(x):
    return int(x * 256)   # Q8.8 format
fixed = [to_fixed(x) for x in norm]
print(fixed)
```

```python
for i,v in enumerate(fixed,1):
    if v<0:
        print(f"x{i} = -16'sd{-v};")
    else:
        print(f"x{i} = 16'sd{v};")
```
