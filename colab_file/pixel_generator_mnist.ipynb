```python
from PIL import Image
import numpy as np
img=Image.open("image.png").convert("L")
img=img.resize((28,28))
pixels=np.array(img)
with open("image.mem","w") as f:
    for row in pixels:
        for p in row:
            f.write(f"{p:02X}\n")
print("image.mem generated")
```
